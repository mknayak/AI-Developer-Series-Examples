Lambdas -  Single expression anonymous functions.
unlike C# lambdas , python lambdas can not be nested

```python
count_words=lambda text: len(text.split(' '))
input_text = input("Enter text to count words: ")
print(f"Word count: {count_words(input_text)}")
```

```python
responses = [
    "Hello world",
    "This is a test",
    "Python is great",
    "I love programming",
    "AI is the future",
]

sort_asc_1=sorted(responses, key=lambda text: len(text.split(' ')))
sort_asc_2=sorted(responses, key=count_words)
sort_desc=sorted(responses, key=count_words, reverse=True)

print(f"Sorted responses by word count (ascending):{sort_asc_1}")
print(f"Sorted responses by word count (ascending) using count_words:{sort_asc_2}")
print(f"Sorted responses by word count (descending) using count_words:{sort_desc}")
```

```python
ai_responses = [
    {"response": "Hello world", "input_tokens": 2, "output_tokens": 2, "score": 0.9},
    {"response": "This is a test", "input_tokens": 4, "output_tokens": 4, "score": 0.8},
    {"response": "Python is great", "input_tokens": 3, "output_tokens": 3, "score": 0.95},
    {"response": "I love programming", "input_tokens": 3, "output_tokens": 3, "score": 0.85},
    {"response": "AI is the future", "input_tokens": 5, "output_tokens": 5, "score": 0.9},
]

print(f"AI Responses sorted by score (descending): {sorted(ai_responses, key=lambda x: x['score'], reverse=True)}")
print(f"AI Responses filtered by score (>=0.9): {list(filter(lambda x: x['score']>=0.9, ai_responses))}")
print(f"AI Responses filtered by tokens (>=6): {list(filter(lambda x: (x['input_tokens']+x['output_tokens'])>=6, ai_responses))}")
```

**Higher Order Functions**
Functions can be passed as arguments, return from another functions or store them in a list. Like C# delegates or Func<>


```python
from typing import Callable

ai_responses = [
    {"response": "Hello world", "input_tokens": 2, "output_tokens": 2, "score": 0.9},
    {"response": "This is a test", "input_tokens": 4, "output_tokens": 4, "score": 0.8},
    {"response": "Python is great", "input_tokens": 3, "output_tokens": 3, "score": 0.95},
    {"response": "I love programming", "input_tokens": 3, "output_tokens": 3, "score": 0.85},
    {"response": "AI is the future", "input_tokens": 5, "output_tokens": 5, "score": 0.9},
]
def filter_response(responses:list[dict], filter_func:Callable[[dict], bool]) -> None:
    new_response = [r["response"] for r in responses if filter_func(r)] # comprehension : [r for r in list if condition]
    print(f"Filtered responses: {new_response}")
    
filter_response(ai_responses, lambda x: x['score']>=0.9)

def filter_token(obj:dict, min_tokens:int) -> bool:
    return (obj['input_tokens'] + obj['output_tokens']) >= min_tokens

filter_response(ai_responses, lambda x: filter_token(x, 6))
filter_response(ai_responses, lambda x: filter_token(x, 8))
```
